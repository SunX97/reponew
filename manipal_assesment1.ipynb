```python
#Assesment 1 Find the temperature data of my town
import numpy as np
temperatures = np.array([31.2, 32.5, 35.1,36.0,34.2,33.1,30.5,29.8,31.0,35.5,37.7,36.8,34.5,32.0,31.5,33.2,36.1, 35.1, 34.0, 32.8,30.2,31.5,36.5,37.0,35.0,33.0,32.1,31.8,34.0,36.3,32.2])
#1. FInd the hottest and coldest temperatures
hottest_temp = np.max(temperatures)
coldest_temp = np.min(temperatures)
#1.1 Find the exact days (adding 1 because array indices start at 0)
hottest_day = np.where(temperatures == hottest_temp)[0][0] + 1
coldest_day = np.where(temperatures == coldest_temp)[0][0] + 1

#2. Find the average temperature of the month
avg_temp = np.mean(temperatures)

#3. Find and list all days where temperatures were above 35
days_above_35 = np.where(temperatures > 35)[0] + 1

print("Hottest temperature:", hottest_temp, "on day", hottest_day)
print("Coldest temperature:", coldest_temp, "on day", coldest_day)
print("Average temperature:", avg_temp)
print("Days with temperatures above 35:", days_above_35)
```
