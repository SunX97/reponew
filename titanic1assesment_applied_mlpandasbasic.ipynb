```python
import pandas as pd
import zipfile

df=pd.read_csv('/content/Titanic-Dataset.csv')
df.head()
```

```python
#task 2 show total number of passengers
total_passengers = len(df)
print(f"Total number of passengers: {total_passengers}")
```

```python
avg_age_survivors =df[df['Survived']==1]['Age'].mean()
print(f"Average age of survivors: {avg_age_survivors:.2f}")
```

```python
avg_age_of_non_survivors = df[df['Survived']==0]['Age'].mean()
print(f"Average age of non survivors: {avg_age_of_non_survivors:.2f}")
```

```python
# List the passengers older than 60 years or abovbe
older_than_sixty = df[df['Age'] > 60]
print(f"\n--- Passengers older than 60 years ---")
print(older_than_sixty)
```

```python
# Find the average fare money for each passenger class
avg_fare_by_class = df.groupby('Pclass')['Fare'].mean()
print("\n--- Average Fare by Passenger Class  ---")
print(avg_fare_by_class)
```
