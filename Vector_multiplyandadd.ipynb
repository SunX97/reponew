```python
%%writefile vector_module.py
import sys
import random
import math
from typing import Self

class Vec:
    def __init__(self, src=None) -> Self:
        src_list = list(src) if src is not None else []
        if any(not isinstance(x, (int, float)) for x in src_list):
            raise TypeError("All vector elements must be numbers")
        self.elements = src_list

    def _check_vec(self, t: Self):
        """User-defined helper to validate type and dimensions for vector-vector operations."""
        if not isinstance(t, Vec):
            raise TypeError(f"Expected Vec, got {type(t)}")
        if len(self.elements) != len(t.elements):
            raise TypeError("Vectors must have identical dimensions")

    def __add__(self, t: Self) -> Self:
        # Applied here
        self._check_vec(t)
        return Vec([round(x + y, 5) for x, y in zip(self.elements, t.elements)])

    def __sub__(self, t: Self) -> Self:
        # Applied here
        self._check_vec(t)
        return Vec([round(x - y, 5) for x, y in zip(self.elements, t.elements)])

    def __iadd__(self, t: Self) -> Self:
        # Applied here
        self._check_vec(t)
        for i in range(len(self.elements)):
            self.elements[i] = round(self.elements[i] + t.elements[i], 5)
        return self

    def __radd__(self, other) -> Self:
        # Applied here via type checking link
        if not isinstance(other, Vec):
            raise TypeError(f"Addition with non-Vec type {type(other)} is not supported")
        self._check_vec(other)
        return other.__add__(self)

    def __neg__(self) -> Self:
        return Vec([round(-x, 5) for x in self.elements])

    def __rmul__(self, scalar: int | float) -> Self:
        if not isinstance(scalar, (int, float)):
            raise TypeError("Scalar must be an int or float")
        return Vec([round(x * scalar, 5) for x in self.elements])

    def __imul__(self, scalar: int | float) -> Self:
        if not isinstance(scalar, (int, float)):
            raise TypeError("Scalar must be an int or float")
        for i in range(len(self.elements)):
            self.elements[i] = round(self.elements[i] * scalar, 5)
        return self

    def __repr__(self) -> str: return repr(self.elements)
    def __len__(self) -> int: return len(self.elements)

    @staticmethod
    def zeros(n: int) -> Self:
        if n <= 0: raise ValueError("Dimension must be > 0")
        return Vec([0.0] * n)

    @staticmethod
    def ones(n: int) -> Self:
        if n <= 0: raise ValueError("Dimension must be > 0")
        return Vec([1.0] * n)

    @staticmethod
    def uniform(n: int) -> Self:
        if n <= 0: raise ValueError("Dimension must be > 0")
        return Vec([random.random() for _ in range(n)])

    def norm(self) -> float:
        return round(math.sqrt(sum(x ** 2 for x in self.elements)), 5)
```

```python
%%writefile test_vector.py
from vector_module import Vec

def run_tests():
    print("--- Running Unit Tests ---")
    v1 = Vec([1.0, 2.0, 3.0])
    v2 = Vec([4.0, 5.0, 6.0])

    assert str(v1 + v2) == "[5.0, 7.0, 9.0]", "Addition failed"
    assert str(v2 - v1) == "[3.0, 3.0, 3.0]", "Subtraction failed"
    assert str(-v1) == "[-1.0, -2.0, -3.0]", "Negation failed"
    assert str(2.0 * v1) == "[2.0, 4.0, 6.0]", "Scalar multiplication failed"

    v_norm = Vec([3.0, 4.0])
    assert v_norm.norm() == 5.0, "Norm computation failed"

    print("All unit tests passed successfully!")

if __name__ == "__main__":
    run_tests()
```

```python
%%writefile benchmark.py
import time
import numpy as np
from vector_module import Vec

def run_benchmarks():
    print("\n--- Running Performance Benchmarks ---")
    dimensions = [2048, 4096, 8192, 16384, 32768, 65536]

    print(f"{'Dim':<8} | {'Custom Add (s)':<15} | {'NumPy Add (s)':<15} | {'Custom Norm (s)':<15} | {'NumPy Norm (s)':<15}")
    print("-" * 78)

    for n in dimensions:
        py_v1 = Vec.uniform(n)
        py_v2 = Vec.uniform(n)
        np_v1 = np.array(py_v1.elements)
        np_v2 = np.array(py_v2.elements)

        # Custom Addition
        start = time.perf_counter()
        for _ in range(10):
            _ = py_v1 + py_v2
        custom_add = (time.perf_counter() - start) / 10

        # NumPy Addition
        start = time.perf_counter()
        for _ in range(10):
            _ = np_v1 + np_v2
        numpy_add = (time.perf_counter() - start) / 10

        # Custom Norm
        start = time.perf_counter()
        for _ in range(10):
            _ = py_v1.norm()
        custom_norm = (time.perf_counter() - start) / 10

        # NumPy Norm
        start = time.perf_counter()
        for _ in range(10):
            _ = np.linalg.norm(np_v1)
        numpy_norm = (time.perf_counter() - start) / 10

        print(f"{n:<8} | {custom_add:<15.6f} | {numpy_add:<15.6f} | {custom_norm:<15.6f} | {numpy_norm:<15.6f}")

if __name__ == "__main__":
    run_benchmarks()
```
