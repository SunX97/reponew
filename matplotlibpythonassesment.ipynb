```python
from sklearn.datasets import load_iris
iris = load_iris(as_frame = True)
df = iris.frame
df.head()
```

```python
#Activity 2
plt.scatter(df["sepal length (cm)"],df["petal length (cm)"])
plt.xlabel("sepal length (cm)")
plt.ylabel("sepal width (cm)")
plt.show()
```

```python
#Activity 3
plt.hist(df["petal width (cm)"])
plt.title("Distribution")
plt.show()
```

```python
import matplotlib.pyplot as plt

# Days of the month (July has 31 days)
days = list(range(1, 32))

# Real historical daily high temperatures for July 2026 in Celcius
# Source: Historical Weather Data for Dhrangadhra & Udupi
dhrangadhra_highs = [32,31,30,31,31,30,28,32,32,32,32,32,32,31,30,
             31,31,31,31,30,29,30,27,26,27,27,29,30,30,30,29]

udupi_highs = [25,25,25,24,25,25,25,25,26,27,27,27,27,27,27,26,
               25,26,26,25,26,24,25,25,25,26,25,26,25,25,24]


# Plot lines with distinct colors, markers, and custom labels
plt.plot(days, dhrangadhra_highs, color='green', marker='o', label='Dhrangadhra')
plt.plot(days, udupi_highs, color='blue', marker='s', label='Udupi')

# Add titles and axis labels
plt.title('Daily High Temperatures - July 2026', fontsize=14, fontweight='bold')
plt.xlabel('Day of July', fontsize=12)
plt.ylabel('Temperature (°C)', fontsize=12)

plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(fontsize=11, loc='best')
plt.legend()

# Show the plot

plt.show()
```
